```python
#Merging data bases
#I want a data base with COLLISION_ID, BOROUGH and STREET NAME (because I noticed sometimes the Borough is not listed)
```

```python
#From all of the columns of collisions_nyc, I want to keep COLLISION_ID, BOROUGH and STREET NAME
location_cols = [
    "COLLISION_ID",
    "BOROUGH",
    "ON STREET NAME"
]
```

```python
#I create an auxiliary table that I set aside for the moment, which allows me to keep the original collisions_nyc base intact.
location_lookup = collisions_nyc[location_cols].copy()
```

```python
#Merge (through COLLISION_ID because they share it)
escooters_geo = crash_escooters.merge(
    location_lookup,
    on="COLLISION_ID",
    how="left",
    validate="many_to_one"
)
```

```python
#My new data base has borough and street name now!
escooters_geo.head()
```

```python
#Which borough shows up the most?
escooters_geo.groupby("BOROUGH")["COLLISION_ID"].nunique().sort_values(ascending=False)
```

```python
escooters_geo["BOROUGH"].isna().sum()
```
